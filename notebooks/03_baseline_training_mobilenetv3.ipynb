```python
from pathlib import Path
import shutil
import hashlib
import time

# --------------------------------------------------
# Source and persistent destination
# --------------------------------------------------
source_cache = Path("/content/fruit_freshness_cache")
source_images = source_cache / "images"
source_manifest = source_cache / "train_validation_cache_manifest.csv"

project_root = Path("/content/drive/MyDrive/Fruit_Freshness_Research")
persistent_cache = project_root / "datasets" / "cache" / "train_validation_cache"
persistent_images = persistent_cache / "images"
persistent_manifest = persistent_cache / "train_validation_cache_manifest.csv"

# --------------------------------------------------
# Safety checks
# --------------------------------------------------
assert source_images.exists(), "Source image cache is missing!"
assert source_manifest.exists(), "Source cache manifest is missing!"

source_files = sorted(
    p for p in source_images.iterdir()
    if p.is_file()
)

print(f"Source image files: {len(source_files)}")

# Create persistent destination folders
persistent_images.mkdir(parents=True, exist_ok=True)

# --------------------------------------------------
# Copy (resume-safe: skip only same-size files)
# --------------------------------------------------
print("\nCopying images to Google Drive...")
start_time = time.time()

copied = 0
skipped = 0
copy_errors = []

for i, src in enumerate(source_files, start=1):
    dst = persistent_images / src.name

    try:
        # Skip only if the existing file has the same size
        if dst.exists() and dst.stat().st_size == src.stat().st_size:
            skipped += 1
        else:
            shutil.copy2(src, dst)
            copied += 1

        if i % 250 == 0 or i == len(source_files):
            print(
                f"Progress: {i}/{len(source_files)} | "
                f"copied={copied} | skipped={skipped} | "
                f"errors={len(copy_errors)}"
            )

    except Exception as e:
        copy_errors.append((src.name, str(e)))

# Copy cache manifest
shutil.copy2(source_manifest, persistent_manifest)

# --------------------------------------------------
# Verification: count and size
# --------------------------------------------------
destination_files = sorted(
    p for p in persistent_images.iterdir()
    if p.is_file()
)

source_total_bytes = sum(p.stat().st_size for p in source_files)
destination_total_bytes = sum(p.stat().st_size for p in destination_files)

source_map = {p.name: p.stat().st_size for p in source_files}
destination_map = {p.name: p.stat().st_size for p in destination_files}

missing_files = sorted(set(source_map) - set(destination_map))
extra_files = sorted(set(destination_map) - set(source_map))

size_mismatches = [
    name for name in source_map
    if name in destination_map
    and source_map[name] != destination_map[name]
]

# --------------------------------------------------
# SHA-256 verification
# Files are named by their SHA-256 content hash.
# Verify Drive copies against their filename.
# --------------------------------------------------
print("\nVerifying SHA-256 hashes of Drive images...")

hash_verified = 0
hash_failed = []

for i, file_path in enumerate(destination_files, start=1):
    expected_hash = file_path.stem.lower()

    try:
        h = hashlib.sha256()

        with open(file_path, "rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                h.update(chunk)

        actual_hash = h.hexdigest()

        if actual_hash == expected_hash:
            hash_verified += 1
        else:
            hash_failed.append(
                (file_path.name, expected_hash, actual_hash)
            )

        if i % 250 == 0 or i == len(destination_files):
            print(
                f"Hash progress: {i}/{len(destination_files)} | "
                f"verified={hash_verified} | failed={len(hash_failed)}"
            )

    except Exception as e:
        hash_failed.append((file_path.name, str(e)))

# --------------------------------------------------
# Final report
# --------------------------------------------------
elapsed = time.time() - start_time

checks = {
    "Expected source count (4133)": len(source_files) == 4133,
    "Destination count matches source": len(destination_files) == len(source_files),
    "No missing files": len(missing_files) == 0,
    "No extra files": len(extra_files) == 0,
    "No size mismatches": len(size_mismatches) == 0,
    "All destination hashes verified": len(hash_failed) == 0
        and hash_verified == len(destination_files),
    "Manifest copied": persistent_manifest.exists(),
    "No copy errors": len(copy_errors) == 0,
}

print("\n" + "=" * 70)
print("PERSISTENT CACHE COPY REPORT")
print("=" * 70)
print(f"Source images:       {len(source_files)}")
print(f"Destination images:   {len(destination_files)}")
print(f"Copied this run:     {copied}")
print(f"Skipped existing:    {skipped}")
print(f"SHA-256 verified:    {hash_verified}")
print(f"Hash failures:       {len(hash_failed)}")
print(f"Copy errors:         {len(copy_errors)}")
print(f"Missing files:       {len(missing_files)}")
print(f"Extra files:         {len(extra_files)}")
print(f"Size mismatches:     {len(size_mismatches)}")
print(f"Source size:         {source_total_bytes / (1024**3):.2f} GB")
print(f"Drive size:          {destination_total_bytes / (1024**3):.2f} GB")
print(f"Elapsed time:        {elapsed / 60:.1f} minutes")

print("\nPersistent destination:")
print(persistent_cache)

print("\n" + "=" * 70)
print("FINAL VERDICT")
print("=" * 70)

for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'} | {name}")

print(f"\nOverall: {'PASS' if all(checks.values()) else 'FAIL — investigate'}")
print("Test split not accessed.")
```

```python

from pathlib import Path
import pandas as pd
import zipfile
import hashlib
import shutil
import os
import time

# ============================================================
# PATHS
# ============================================================
ROOT = Path("/content/drive/MyDrive/Fruit_Freshness_Research")

MANIFEST_PATH = (
    ROOT / "datasets/audit/quality_control/"
    "final_dataset_checkpoint/final_record_split_manifest_v2.csv"
)

AFRUITDB_ZIP = ROOT / "datasets/original/AFruitDB_original.zip"
EXPANDED_ZIP = ROOT / "datasets/original/Bangladeshi_Fruit_Grade_original.zip"

# Temporary Colab cache
LOCAL_CACHE = Path("/content/fruit_freshness_cache")
LOCAL_IMAGES = LOCAL_CACHE / "images"
LOCAL_CACHE_MANIFEST = LOCAL_CACHE / "train_validation_cache_manifest.csv"

# Persistent Drive cache
DRIVE_CACHE = ROOT / "datasets/cache/train_validation_cache"
DRIVE_IMAGES = DRIVE_CACHE / "images"
DRIVE_MANIFEST = DRIVE_CACHE / "train_validation_cache_manifest.csv"

LOCAL_CACHE.mkdir(parents=True, exist_ok=True)
LOCAL_IMAGES.mkdir(parents=True, exist_ok=True)
DRIVE_IMAGES.mkdir(parents=True, exist_ok=True)

# ============================================================
# LOAD FROZEN MANIFEST — TRAIN AND VALIDATION ONLY
# ============================================================
manifest = pd.read_csv(MANIFEST_PATH)

train_val = manifest[
    manifest["split"].isin(["train", "validation"])
].copy()

unique_images = train_val.drop_duplicates(subset="sha256").copy()

assert len(unique_images) == 4133, (
    f"Expected 4133 unique train/validation images, got {len(unique_images)}"
)

# ============================================================
# HASH HELPERS
# ============================================================
def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

def sha256_bytes(data):
    return hashlib.sha256(data).hexdigest()

# ============================================================
# CACHE RECONSTRUCTION + DRIVE PERSISTENCE
# ============================================================
print("=" * 70)
print("CACHE REBUILD + DRIVE PERSISTENCE")
print("=" * 70)

new_local = 0
reused_local = 0
repaired_local = 0
new_drive = 0
reused_drive = 0
failures = []

temporary_nested_zip = LOCAL_CACHE / "AFruitDB_inner_temporary.zip"
start_time = time.time()

try:
    # --------------------------------------------------------
    # Prepare nested AFruitDB ZIP on local disk
    # --------------------------------------------------------
    print("\n[1] Preparing temporary AFruitDB inner ZIP...")

    with zipfile.ZipFile(AFRUITDB_ZIP, "r") as outer:
        nested_name = next(
            name for name in outer.namelist()
            if name.lower().endswith(".zip")
        )

        with outer.open(nested_name) as src:
            with open(temporary_nested_zip, "wb") as dst:
                shutil.copyfileobj(src, dst, length=1024 * 1024)

    print("Temporary nested ZIP prepared.")

    # --------------------------------------------------------
    # Open source archives and reconstruct/cache images
    # --------------------------------------------------------
    with zipfile.ZipFile(temporary_nested_zip, "r") as afruit_inner, \
         zipfile.ZipFile(EXPANDED_ZIP, "r") as expanded_archive:

        for i, (_, row) in enumerate(unique_images.iterrows(), start=1):

            expected_hash = str(row["sha256"])
            relative_path = str(row["relative_path"]).replace("\\", "/")
            dataset = row["dataset"]

            local_path = LOCAL_IMAGES / f"{expected_hash}.jpg"
            drive_path = DRIVE_IMAGES / f"{expected_hash}.jpg"

            # ------------------------------------------------
            # 1. Verify or reconstruct local image
            # ------------------------------------------------
            local_valid = (
                local_path.exists()
                and sha256_file(local_path) == expected_hash
            )

            if local_valid:
                reused_local += 1
            else:
                if local_path.exists():
                    local_path.unlink()
                    repaired_local += 1

                # Read from the correct original archive
                if dataset == "AFruitDB":
                    image_bytes = afruit_inner.read(relative_path)
                elif dataset == "Expanded":
                    image_bytes = expanded_archive.read(relative_path)
                else:
                    raise ValueError(f"Unexpected dataset: {dataset}")

                # Verify bytes before writing
                if sha256_bytes(image_bytes) != expected_hash:
                    raise ValueError(
                        f"Source SHA-256 mismatch for {relative_path}"
                    )

                local_temp = local_path.with_suffix(".tmp")
                with open(local_temp, "wb") as f:
                    f.write(image_bytes)

                # Verify written local file
                if sha256_file(local_temp) != expected_hash:
                    local_temp.unlink(missing_ok=True)
                    raise ValueError(
                        f"Local verification failed: {expected_hash}"
                    )

                os.replace(local_temp, local_path)
                new_local += 1

            # ------------------------------------------------
            # 2. Persist image to Drive
            # ------------------------------------------------
            drive_valid = (
                drive_path.exists()
                and sha256_file(drive_path) == expected_hash
            )

            if drive_valid:
                reused_drive += 1
            else:
                drive_temp = drive_path.with_suffix(".tmp")

                # Copy verified local file to Drive temporary file
                shutil.copy2(local_path, drive_temp)

                # Verify Drive copy before finalizing
                if sha256_file(drive_temp) != expected_hash:
                    drive_temp.unlink(missing_ok=True)
                    raise ValueError(
                        f"Drive-copy SHA-256 verification failed: {expected_hash}"
                    )

                os.replace(drive_temp, drive_path)
                new_drive += 1

            # ------------------------------------------------
            # Progress
            # ------------------------------------------------
            if i % 100 == 0 or i == len(unique_images):
                print(
                    f"Progress {i}/{len(unique_images)} | "
                    f"local_new={new_local} | "
                    f"local_reused={reused_local} | "
                    f"drive_new={new_drive} | "
                    f"drive_reused={reused_drive} | "
                    f"errors={len(failures)}"
                )

except Exception as e:
    failures.append(str(e))
    print("\nERROR:", repr(e))

finally:
    # Remove nested temporary ZIP
    if temporary_nested_zip.exists():
        temporary_nested_zip.unlink()

# ============================================================
# BUILD PERSISTENT MANIFEST
# ============================================================
persistent_manifest = unique_images[
    ["sha256", "dataset", "relative_path", "fruit", "grade", "split"]
].copy()

persistent_manifest["cache_path"] = persistent_manifest["sha256"].apply(
    lambda h: str(DRIVE_IMAGES / f"{h}.jpg")
)

persistent_manifest["cache_exists"] = persistent_manifest["cache_path"].apply(
    lambda p: Path(p).exists()
)

persistent_manifest["cache_verified"] = persistent_manifest.apply(
    lambda r: (
        r["cache_exists"]
        and sha256_file(Path(r["cache_path"])) == r["sha256"]
    ),
    axis=1
)

persistent_manifest.to_csv(DRIVE_MANIFEST, index=False)

# Also save a local copy of this manifest
persistent_manifest.to_csv(LOCAL_CACHE_MANIFEST, index=False)

# ============================================================
# FINAL REPORT
# ============================================================
drive_files = [
    p for p in DRIVE_IMAGES.glob("*.jpg")
    if p.is_file()
]

drive_bytes = sum(p.stat().st_size for p in drive_files)
verified_count = int(persistent_manifest["cache_verified"].sum())

checks = {
    "Unique train/validation count is 4133":
        len(unique_images) == 4133,

    "Drive image count is 4133":
        len(drive_files) == 4133,

    "All 4133 Drive images verified":
        verified_count == 4133,

    "Persistent manifest saved":
        DRIVE_MANIFEST.exists(),

    "Persistent manifest rows correct":
        len(persistent_manifest) == 4133,

    "No cache reconstruction/copy errors":
        len(failures) == 0,
}

print("\n" + "=" * 70)
print("FINAL CACHE + DRIVE PERSISTENCE REPORT")
print("=" * 70)

print(f"Images requested:           {len(unique_images)}")
print(f"New local files:           {new_local}")
print(f"Reused valid local files:  {reused_local}")
print(f"Repaired local files:      {repaired_local}")
print(f"New Drive files:           {new_drive}")
print(f"Reused valid Drive files:  {reused_drive}")
print(f"Drive image files:          {len(drive_files)}")
print(f"Drive SHA verified:         {verified_count}")
print(f"Drive cache size:          {drive_bytes / (1024**3):.2f} GB")
print(f"Failures:                  {len(failures)}")
print(f"Elapsed:                   {(time.time()-start_time)/60:.1f} minutes")

print("\nPersistent cache location:")
print(DRIVE_CACHE)

print("\nPersistent manifest:")
print(DRIVE_MANIFEST)

print("\n" + "=" * 70)
print("CHECK RESULTS")
print("=" * 70)

for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'} | {name}")

print("\nOVERALL:", "PASS" if all(checks.values()) else "REVIEW REQUIRED")
print("Test split not accessed.")
```

```python

# ============================================================
# DATASET + DATALOADER PREFLIGHT
# Train and validation only — no test data accessed
# ============================================================

from pathlib import Path
import json
import random
import numpy as np
import pandas as pd
from PIL import Image, ImageFile

import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

# -----------------------------
# 1. Reproducibility
# -----------------------------
SEED = 20261119

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

ImageFile.LOAD_TRUNCATED_IMAGES = False

# -----------------------------
# 2. Paths
# -----------------------------
PROJECT_ROOT = Path("/content/drive/MyDrive/Fruit_Freshness_Research")
EXPERIMENT_DIR = PROJECT_ROOT / "experiments" / "baseline_mobilenetv3_small"

TRAIN_MANIFEST = EXPERIMENT_DIR / "unique_train_manifest.csv"
VAL_MANIFEST = EXPERIMENT_DIR / "unique_validation_manifest.csv"
CACHE_MANIFEST = (
    PROJECT_ROOT
    / "datasets/cache/train_validation_cache"
    / "train_validation_cache_manifest.csv"
)
CLASS_MAP_PATH = EXPERIMENT_DIR / "class_to_index.json"

# -----------------------------
# 3. Load frozen manifests
# -----------------------------
train_df = pd.read_csv(TRAIN_MANIFEST)
val_df = pd.read_csv(VAL_MANIFEST)
cache_df = pd.read_csv(CACHE_MANIFEST)

with open(CLASS_MAP_PATH, "r") as f:
    class_to_index = json.load(f)

# Normalize mapping values to integers
class_to_index = {
    str(name): int(index)
    for name, index in class_to_index.items()
}

index_to_class = {
    index: name for name, index in class_to_index.items()
}

# -----------------------------
# 4. Exact SHA/cache reconciliation
# -----------------------------
required_manifest_columns = {"sha256", "class_name"}
required_cache_columns = {"sha256", "split", "cache_path"}

for label, df in [
    ("train", train_df),
    ("validation", val_df),
]:
    missing = required_manifest_columns - set(df.columns)
    assert not missing, f"{label} manifest missing columns: {missing}"

missing = required_cache_columns - set(cache_df.columns)
assert not missing, f"Cache manifest missing columns: {missing}"

train_hashes = set(train_df["sha256"].astype(str))
val_hashes = set(val_df["sha256"].astype(str))
cache_hashes = set(cache_df["sha256"].astype(str))

expected_hashes = train_hashes | val_hashes

assert len(train_hashes) == len(train_df), "Duplicate train SHA values"
assert len(val_hashes) == len(val_df), "Duplicate validation SHA values"
assert train_hashes.isdisjoint(val_hashes), "Train/validation SHA overlap"
assert cache_hashes == expected_hashes, (
    "Cache SHA set does not exactly match train + validation SHA sets"
)

assert len(class_to_index) == 21, "Expected 21 classes"
assert sorted(class_to_index.values()) == list(range(21)), (
    "Class indices must be contiguous from 0 to 20"
)

cache_lookup = cache_df.set_index("sha256").to_dict("index")

for split_name, df in [
    ("train", train_df),
    ("validation", val_df),
]:
    for class_name in df["class_name"].astype(str).unique():
        assert class_name in class_to_index, (
            f"Unknown class in {split_name}: {class_name}"
        )

    for sha in df["sha256"].astype(str):
        assert sha in cache_lookup, f"Missing cache SHA: {sha}"
        assert cache_lookup[sha]["split"] == split_name, (
            f"Incorrect cache split for SHA {sha}"
        )
        assert Path(cache_lookup[sha]["cache_path"]).is_file(), (
            f"Missing image file for SHA {sha}"
        )

# -----------------------------
# 5. Image transformations
# -----------------------------
# ImageNet normalization is used for the pretrained MobileNetV3 model.
IMAGE_SIZE = 224

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),
])

val_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),
])

# -----------------------------
# 6. Dataset class
# -----------------------------
class FruitFreshnessDataset(Dataset):
    def __init__(self, dataframe, cache_lookup, class_mapping, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.cache_lookup = cache_lookup
        self.class_mapping = class_mapping
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        sha = str(row["sha256"])
        class_name = str(row["class_name"])

        image_path = Path(self.cache_lookup[sha]["cache_path"])

        with Image.open(image_path) as image:
            image = image.convert("RGB")

            if self.transform is not None:
                image = self.transform(image)

        label = self.class_mapping[class_name]

        return image, torch.tensor(label, dtype=torch.long)

# -----------------------------
# 7. Build datasets and loaders
# -----------------------------
train_dataset = FruitFreshnessDataset(
    train_df, cache_lookup, class_to_index, train_transform
)

val_dataset = FruitFreshnessDataset(
    val_df, cache_lookup, class_to_index, val_transform
)

BATCH_SIZE = 32

train_generator = torch.Generator()
train_generator.manual_seed(SEED)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
    generator=train_generator,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
)

# -----------------------------
# 8. Inspect first batch
# -----------------------------
train_images, train_labels = next(iter(train_loader))
val_images, val_labels = next(iter(val_loader))

assert train_images.shape == (BATCH_SIZE, 3, 224, 224)
assert val_images.shape == (BATCH_SIZE, 3, 224, 224)

assert train_labels.dtype == torch.long
assert val_labels.dtype == torch.long

assert train_labels.min().item() >= 0
assert train_labels.max().item() < 21
assert val_labels.min().item() >= 0
assert val_labels.max().item() < 21

# -----------------------------
# 9. Decode every image once
# -----------------------------
# This checks the full train/validation image collections,
# rather than only the first batches.
decode_errors = []

for split_name, dataframe in [
    ("train", train_df),
    ("validation", val_df),
]:
    for sha in dataframe["sha256"].astype(str):
        path = Path(cache_lookup[sha]["cache_path"])
        try:
            with Image.open(path) as image:
                image.verify()

            with Image.open(path) as image:
                image.convert("RGB").load()

        except Exception as exc:
            decode_errors.append((split_name, sha, str(exc)))

# -----------------------------
# 10. Report
# -----------------------------
print("=" * 65)
print("DATASET + DATALOADER PREFLIGHT")
print("=" * 65)

print(f"Train dataset images:       {len(train_dataset)}")
print(f"Validation dataset images:  {len(val_dataset)}")
print(f"Combined images:            {len(train_dataset) + len(val_dataset)}")
print(f"Number of classes:          {len(class_to_index)}")
print(f"Batch size:                 {BATCH_SIZE}")
print(f"Input image shape:          {tuple(train_images.shape)}")
print(f"Validation batch shape:     {tuple(val_images.shape)}")
print(f"Train label dtype:          {train_labels.dtype}")
print(f"Validation label dtype:     {val_labels.dtype}")
print(f"Train label range:          {train_labels.min().item()}–{train_labels.max().item()}")
print(f"Validation label range:     {val_labels.min().item()}–{val_labels.max().item()}")
print(f"Image decode errors:        {len(decode_errors)}")

print("\nDevice available:")
print(torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU")

print("\nChecks:")
checks = {
    "Exact cache SHA reconciliation": cache_hashes == expected_hashes,
    "Train dataset size": len(train_dataset) == 3401,
    "Validation dataset size": len(val_dataset) == 732,
    "21-class mapping": len(class_to_index) == 21,
    "Train batch shape": tuple(train_images.shape) == (32, 3, 224, 224),
    "Validation batch shape": tuple(val_images.shape) == (32, 3, 224, 224),
    "Valid train labels": train_labels.min().item() >= 0 and train_labels.max().item() < 21,
    "Valid validation labels": val_labels.min().item() >= 0 and val_labels.max().item() < 21,
    "All images decode": len(decode_errors) == 0,
}

for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'} | {name}")

print("\nOVERALL:", "PASS" if all(checks.values()) else "FAIL")

if decode_errors:
    print("\nFirst decode errors:")
    for error in decode_errors[:10]:
        print(error)

print("\nPreflight only. No model training started.")
print("Test split not accessed.")
```

```python

# ============================================================
# TRAINING SCRIPT PREPARATION (VERSION 1)
# Writes script + syntax-checks only. DOES NOT TRAIN.
# ============================================================

from pathlib import Path
import py_compile
import importlib.util

PROJECT_ROOT = Path("/content/drive/MyDrive/Fruit_Freshness_Research")
EXPERIMENT_DIR = PROJECT_ROOT / "experiments" / "baseline_mobilenetv3_small"

SCRIPT_PATH = EXPERIMENT_DIR / "train_baseline_v1.py"

assert (EXPERIMENT_DIR / "resolved_training_config_v1.json").is_file(), (
    "Resolved config not found."
)
assert (EXPERIMENT_DIR / "unique_train_manifest.csv").is_file()
assert (EXPERIMENT_DIR / "unique_validation_manifest.csv").is_file()
assert (
    PROJECT_ROOT
    / "datasets/cache/train_validation_cache/train_validation_cache_manifest.csv"
).is_file()

if SCRIPT_PATH.exists():
    print("WARNING: Training script already exists; refusing to overwrite.")
    print("Existing script:", SCRIPT_PATH)
else:

    script_source = r'''
# ============================================================
# Fruit Freshness Baseline Training
# MobileNetV3-Small | 21 classes
#
# IMPORTANT:
# - Train and validation only.
# - Test split is never loaded by this script.
# - Supports resume from latest checkpoint.
# - This script is not automatically launched when created.
# ============================================================

from pathlib import Path
import json
import random
import csv
import argparse
import os
import tempfile

import numpy as np
import pandas as pd
from PIL import Image, ImageFile

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.models import (
    mobilenet_v3_small,
    MobileNet_V3_Small_Weights,
)

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
)

# ------------------------------------------------------------
# Paths and constants
# ------------------------------------------------------------
PROJECT_ROOT = Path("/content/drive/MyDrive/Fruit_Freshness_Research")
EXPERIMENT_DIR = PROJECT_ROOT / "experiments" / "baseline_mobilenetv3_small"

CONFIG_PATH = EXPERIMENT_DIR / "resolved_training_config_v1.json"
TRAIN_MANIFEST = EXPERIMENT_DIR / "unique_train_manifest.csv"
VAL_MANIFEST = EXPERIMENT_DIR / "unique_validation_manifest.csv"
CLASS_MAP_PATH = EXPERIMENT_DIR / "class_to_index.json"

CACHE_DIR = PROJECT_ROOT / "datasets" / "cache" / "train_validation_cache"
CACHE_MANIFEST = CACHE_DIR / "train_validation_cache_manifest.csv"

CHECKPOINT_DIR = EXPERIMENT_DIR / "checkpoints"
LOG_DIR = EXPERIMENT_DIR / "logs"

BEST_CHECKPOINT = CHECKPOINT_DIR / "best_validation_macro_f1.pt"
LATEST_CHECKPOINT = CHECKPOINT_DIR / "latest_checkpoint.pt"
METRICS_CSV = LOG_DIR / "training_metrics.csv"
PER_CLASS_CSV = LOG_DIR / "per_class_metrics.csv"

ImageFile.LOAD_TRUNCATED_IMAGES = False

# ------------------------------------------------------------
# Reproducibility setup
# ------------------------------------------------------------
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    # Favor repeatability over benchmark speed.
    torch.backends.cudnn.benchmark = False
    torch.backends.cudnn.deterministic = True


# ------------------------------------------------------------
# Dataset
# ------------------------------------------------------------
class FruitDataset(Dataset):
    def __init__(self, dataframe, cache_lookup, class_to_index, transform):
        self.df = dataframe.reset_index(drop=True)
        self.cache_lookup = cache_lookup
        self.class_to_index = class_to_index
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):
        row = self.df.iloc[index]
        sha = str(row["sha256"])
        class_name = str(row["class_name"])

        path = Path(self.cache_lookup[sha]["cache_path"])

        with Image.open(path) as image:
            image = image.convert("RGB")
            image = self.transform(image)

        label = self.class_to_index[class_name]
        return image, torch.tensor(label, dtype=torch.long)


# ------------------------------------------------------------
# Metrics
# ------------------------------------------------------------
def calculate_metrics(y_true, y_pred, class_names):
    accuracy = accuracy_score(y_true, y_pred)

    precision, recall, f1, support = precision_recall_fscore_support(
        y_true,
        y_pred,
        labels=list(range(len(class_names))),
        average=None,
        zero_division=0,
    )

    macro_precision, macro_recall, macro_f1, _ = (
        precision_recall_fscore_support(
            y_true,
            y_pred,
            labels=list(range(len(class_names))),
            average="macro",
            zero_division=0,
        )
    )

    weighted_precision, weighted_recall, weighted_f1, _ = (
        precision_recall_fscore_support(
            y_true,
            y_pred,
            labels=list(range(len(class_names))),
            average="weighted",
            zero_division=0,
        )
    )

    per_class = []
    for i, name in enumerate(class_names):
        per_class.append({
            "class_index": i,
            "class_name": name,
            "precision": float(precision[i]),
            "recall": float(recall[i]),
            "f1": float(f1[i]),
            "support": int(support[i]),
        })

    metrics = {
        "accuracy": float(accuracy),
        "macro_precision": float(macro_precision),
        "macro_recall": float(macro_recall),
        "macro_f1": float(macro_f1),
        "weighted_precision": float(weighted_precision),
        "weighted_recall": float(weighted_recall),
        "weighted_f1": float(weighted_f1),
    }

    return metrics, per_class


# ------------------------------------------------------------
# Epoch functions
# ------------------------------------------------------------
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()

    total_loss = 0.0
    total_samples = 0
    y_true = []
    y_pred = []

    for images, labels in loader:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        logits = model(images)
        loss = criterion(logits, labels)

        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)
        total_loss += loss.item() * batch_size
        total_samples += batch_size

        predictions = logits.argmax(dim=1)

        y_true.extend(labels.detach().cpu().numpy().tolist())
        y_pred.extend(predictions.detach().cpu().numpy().tolist())

    class_names = list(range(21))
    metrics, _ = calculate_metrics(y_true, y_pred, class_names)

    metrics["loss"] = total_loss / max(total_samples, 1)
    return metrics


@torch.no_grad()
def validate_one_epoch(model, loader, criterion, device, class_names):
    model.eval()

    total_loss = 0.0
    total_samples = 0
    y_true = []
    y_pred = []

    for images, labels in loader:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        logits = model(images)
        loss = criterion(logits, labels)

        batch_size = labels.size(0)
        total_loss += loss.item() * batch_size
        total_samples += batch_size

        predictions = logits.argmax(dim=1)

        y_true.extend(labels.cpu().numpy().tolist())
        y_pred.extend(predictions.cpu().numpy().tolist())

    metrics, per_class = calculate_metrics(
        y_true, y_pred, class_names
    )

    metrics["loss"] = total_loss / max(total_samples, 1)
    return metrics, per_class


# ------------------------------------------------------------
# Atomic checkpoint save
# ------------------------------------------------------------
def atomic_save_checkpoint(payload, destination):
    destination.parent.mkdir(parents=True, exist_ok=True)

    # Write to temporary file first, then replace destination.
    # This reduces the risk of a partially written checkpoint.
    with tempfile.NamedTemporaryFile(
        mode="wb",
        dir=destination.parent,
        delete=False,
        suffix=".tmp",
    ) as temp_file:
        temp_path = Path(temp_file.name)

    try:
        torch.save(payload, temp_path)
        os.replace(temp_path, destination)
    finally:
        if temp_path.exists():
            temp_path.unlink()


# ------------------------------------------------------------
# CSV writing
# ------------------------------------------------------------
def write_csv_rows(rows, path):
    if not rows:
        return

    path.parent.mkdir(parents=True, exist_ok=True)

    df = pd.DataFrame(rows)
    df.to_csv(path, index=False)


# ------------------------------------------------------------
# Main
# ------------------------------------------------------------
def main():
    parser = argparse.ArgumentParser()

    parser.add_argument(
        "--resume",
        action="store_true",
        help="Resume from latest checkpoint if available.",
    )

    args = parser.parse_args()

    # Load resolved configuration
    with open(CONFIG_PATH, "r") as f:
        config = json.load(f)

    seed = int(config["reproducibility"]["seed"])
    set_seed(seed)

    device = torch.device(
        "cuda" if torch.cuda.is_available() else "cpu"
    )

    CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)
    LOG_DIR.mkdir(parents=True, exist_ok=True)

    # Load only train and validation manifests.
    train_df = pd.read_csv(TRAIN_MANIFEST)
    val_df = pd.read_csv(VAL_MANIFEST)

    # Load persistent cache index.
    cache_df = pd.read_csv(CACHE_MANIFEST)

    with open(CLASS_MAP_PATH, "r") as f:
        class_to_index = json.load(f)

    class_to_index = {
        str(k): int(v) for k, v in class_to_index.items()
    }

    index_to_class = {
        value: key for key, value in class_to_index.items()
    }

    class_names = [
        index_to_class[i] for i in range(len(index_to_class))
    ]

    # Cache lookup
    cache_lookup = cache_df.set_index("sha256").to_dict("index")

    # Integrity safeguards
    train_hashes = set(train_df["sha256"].astype(str))
    val_hashes = set(val_df["sha256"].astype(str))
    cache_hashes = set(cache_df["sha256"].astype(str))

    assert train_hashes.isdisjoint(val_hashes), (
        "Train/validation SHA overlap detected."
    )

    assert train_hashes | val_hashes == cache_hashes, (
        "Cache SHA set does not match train+validation manifests."
    )

    assert set(train_df["class_name"].astype(str)).issubset(
        set(class_to_index)
    )
    assert set(val_df["class_name"].astype(str)).issubset(
        set(class_to_index)
    )

    # Image transforms
    train_transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(degrees=10),
        transforms.ToTensor(),
        transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
    ])

    val_transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],
        ),
    ])

    train_dataset = FruitDataset(
        train_df, cache_lookup, class_to_index, train_transform
    )
    val_dataset = FruitDataset(
        val_df, cache_lookup, class_to_index, val_transform
    )

    generator = torch.Generator()
    generator.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset,
        batch_size=int(config["training"]["batch_size"]),
        shuffle=True,
        num_workers=0,
        pin_memory=(device.type == "cuda"),
        generator=generator,
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=int(config["training"]["batch_size"]),
        shuffle=False,
        num_workers=0,
        pin_memory=(device.type == "cuda"),
    )

    # Model
    weights = MobileNet_V3_Small_Weights.DEFAULT
    model = mobilenet_v3_small(weights=weights)

    model.classifier[3] = nn.Linear(
        model.classifier[3].in_features,
        len(class_to_index),
    )

    model = model.to(device)

    criterion = nn.CrossEntropyLoss()

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=float(config["training"]["learning_rate"]),
        weight_decay=float(config["training"]["weight_decay"]),
    )

    # Training state
    history = []
    per_class_history = []
    start_epoch = 0
    best_macro_f1 = -1.0
    epochs_without_improvement = 0

    # Resume option
    if args.resume:
        if not LATEST_CHECKPOINT.exists():
            raise FileNotFoundError(
                "Resume requested, but latest checkpoint is missing."
            )

        try:
            checkpoint = torch.load(
                LATEST_CHECKPOINT,
                map_location=device,
                weights_only=False,
            )
        except TypeError:
            checkpoint = torch.load(
                LATEST_CHECKPOINT,
                map_location=device,
            )

        model.load_state_dict(checkpoint["model_state"])
        optimizer.load_state_dict(checkpoint["optimizer_state"])

        start_epoch = int(checkpoint["completed_epoch"])
        best_macro_f1 = float(checkpoint["best_macro_f1"])
        epochs_without_improvement = int(
            checkpoint["epochs_without_improvement"]
        )
        history = checkpoint.get("history", [])
        per_class_history = checkpoint.get("per_class_history", [])

        print(
            f"Resuming after completed epoch {start_epoch}; "
            f"best validation Macro-F1={best_macro_f1:.6f}"
        )

    # Epoch limit
    max_epochs = int(config["training"]["max_epochs"])
    patience = int(config["training"]["early_stopping_patience"])

    # --------------------------------------------------------
    # Training loop
    # --------------------------------------------------------
    for epoch in range(start_epoch, max_epochs):
        epoch_number = epoch + 1

        print("\n" + "=" * 65)
        print(f"Epoch {epoch_number}/{max_epochs}")
        print("=" * 65)

        train_metrics = train_one_epoch(
            model, train_loader, criterion, optimizer, device
        )

        val_metrics, class_metrics = validate_one_epoch(
            model, val_loader, criterion, device, class_names
        )

        row = {
            "epoch": epoch_number,
            "train_loss": train_metrics["loss"],
            "train_accuracy": train_metrics["accuracy"],
            "train_macro_precision": train_metrics["macro_precision"],
            "train_macro_recall": train_metrics["macro_recall"],
            "train_macro_f1": train_metrics["macro_f1"],
            "train_weighted_f1": train_metrics["weighted_f1"],
            "val_loss": val_metrics["loss"],
            "val_accuracy": val_metrics["accuracy"],
            "val_macro_precision": val_metrics["macro_precision"],
            "val_macro_recall": val_metrics["macro_recall"],
            "val_macro_f1": val_metrics["macro_f1"],
            "val_weighted_f1": val_metrics["weighted_f1"],
        }

        history.append(row)

        for item in class_metrics:
            per_class_history.append({
                "epoch": epoch_number,
                **item,
            })

        # Improvement is based on validation Macro-F1 only.
        improved = val_metrics["macro_f1"] > best_macro_f1

        if improved:
            best_macro_f1 = val_metrics["macro_f1"]
            epochs_without_improvement = 0

            best_payload = {
                "completed_epoch": epoch_number,
                "model_state": model.state_dict(),
                "best_macro_f1": best_macro_f1,
                "config_path": str(CONFIG_PATH),
                "seed": seed,
            }

            atomic_save_checkpoint(
                best_payload,
                BEST_CHECKPOINT,
            )
        else:
            epochs_without_improvement += 1

        # Save latest checkpoint after every completed epoch.
        latest_payload = {
            "completed_epoch": epoch_number,
            "model_state": model.state_dict(),
            "optimizer_state": optimizer.state_dict(),
            "best_macro_f1": best_macro_f1,
            "epochs_without_improvement": epochs_without_improvement,
            "history": history,
            "per_class_history": per_class_history,
            "config_path": str(CONFIG_PATH),
            "seed": seed,
        }

        atomic_save_checkpoint(
            latest_payload,
            LATEST_CHECKPOINT,
        )

        # Update CSV logs.
        write_csv_rows(history, METRICS_CSV)
        write_csv_rows(per_class_history, PER_CLASS_CSV)

        print(
            f"Train loss={train_metrics['loss']:.4f} | "
            f"Train Macro-F1={train_metrics['macro_f1']:.4f}"
        )
        print(
            f"Val loss={val_metrics['loss']:.4f} | "
            f"Val accuracy={val_metrics['accuracy']:.4f} | "
            f"Val Macro-F1={val_metrics['macro_f1']:.4f} | "
            f"Val weighted-F1={val_metrics['weighted_f1']:.4f}"
        )
        print(
            f"Best Val Macro-F1={best_macro_f1:.4f} | "
            f"Patience counter={epochs_without_improvement}/{patience}"
        )

        # Early stopping
        if epochs_without_improvement >= patience:
            print(
                "\nEarly stopping triggered: validation Macro-F1 "
                f"did not improve for {patience} consecutive epochs."
            )
            break

    print("\nTraining loop finished.")
    print(f"Best checkpoint: {BEST_CHECKPOINT}")
    print(f"Latest checkpoint: {LATEST_CHECKPOINT}")
    print(f"Metrics CSV: {METRICS_CSV}")
    print(f"Per-class metrics CSV: {PER_CLASS_CSV}")
    print("Test split was not accessed.")


if __name__ == "__main__":
    main()
'''

    SCRIPT_PATH.write_text(script_source, encoding="utf-8")
    print("Training script created:", SCRIPT_PATH)

# ------------------------------------------------------------
# Syntax check — compile without executing
# ------------------------------------------------------------
if SCRIPT_PATH.exists():
    py_compile.compile(str(SCRIPT_PATH), doraise=True)
    print("\nPASS | Training script syntax compiles")

    print("Script size:", SCRIPT_PATH.stat().st_size, "bytes")
    print("Script path:", SCRIPT_PATH)

print("\nNo training launched.")
print("No checkpoint created.")
print("Test split not accessed.")
```

```python
# STEP 5: Add strict checkpoint/resume validation to V3
# This modifies V3 only. It does NOT execute training.

from pathlib import Path
import hashlib
import ast
import os

EXP = Path(
    "/content/drive/MyDrive/Fruit_Freshness_Research/"
    "experiments/baseline_mobilenetv3_small"
)

V3 = EXP / "train_baseline_v3.py"
BACKUP = EXP / "train_baseline_v3_pre_resume_guard.py"

if not V3.exists():
    raise FileNotFoundError("V3 script is missing.")

if BACKUP.exists():
    raise FileExistsError(
        "A V3 pre-guard backup already exists. "
        "No file was overwritten."
    )

original = V3.read_text(encoding="utf-8")
original_hash = hashlib.sha256(V3.read_bytes()).hexdigest()

# Exact insertion point: immediately after checkpoint loading.
anchor = '''        model.load_state_dict(checkpoint["model_state"])
        optimizer.load_state_dict(checkpoint["optimizer_state"])
'''

if original.count(anchor) != 1:
    raise RuntimeError(
        "Expected resume insertion point was not found exactly once. "
        "V3 was not modified."
    )

guard = '''        # --------------------------------------------------------
        # STRICT V3 CHECKPOINT VALIDATION
        # --------------------------------------------------------
        if not isinstance(checkpoint, dict):
            raise RuntimeError("Checkpoint must contain a dictionary.")

        if checkpoint.get("script_version") != "v3":
            raise RuntimeError(
                "Incompatible checkpoint version. "
                "V3 can resume only from checkpoints explicitly "
                "created by V3. V1/V2 checkpoints are not accepted."
            )

        required_fields = [
            "model_state",
            "optimizer_state",
            "completed_epoch",
            "best_macro_f1",
            "epochs_without_improvement",
            "history",
            "per_class_history",
            "python_rng_state",
            "numpy_rng_state",
            "torch_rng_state",
            "loader_generator_state",
        ]

        missing_fields = [
            key for key in required_fields if key not in checkpoint
        ]

        if missing_fields:
            raise RuntimeError(
                "Checkpoint is incomplete. Missing fields: "
                + ", ".join(missing_fields)
            )

        completed_epoch = int(checkpoint["completed_epoch"])
        max_epochs_check = int(config["training"]["max_epochs"])
        patience_check = int(
            config["training"]["early_stopping_patience"]
        )
        patience_counter = int(
            checkpoint["epochs_without_improvement"]
        )

        if completed_epoch < 1:
            raise RuntimeError(
                "Checkpoint completed_epoch must be at least 1."
            )

        if completed_epoch > max_epochs_check:
            raise RuntimeError(
                "Checkpoint epoch exceeds configured maximum epochs."
            )

        if patience_counter < 0:
            raise RuntimeError(
                "Checkpoint patience counter cannot be negative."
            )

        checkpoint_history = checkpoint["history"]

        if not isinstance(checkpoint_history, list):
            raise RuntimeError("Checkpoint history must be a list.")

        history_epochs = [
            int(row["epoch"]) for row in checkpoint_history
        ]

        expected_epochs = list(range(1, completed_epoch + 1))

        if history_epochs != expected_epochs:
            raise RuntimeError(
                "Checkpoint history is inconsistent with completed_epoch."
            )

        if completed_epoch >= max_epochs_check:
            raise RuntimeError(
                "This run has already reached the configured maximum "
                "epoch. Resume is blocked."
            )

        if patience_counter >= patience_check:
            raise RuntimeError(
                "This run has already reached its early-stopping "
                "patience limit. Resume is blocked."
            )

'''

patched = original.replace(anchor, guard + anchor)

# Parse before touching the original.
ast.parse(patched)

# Preserve exact original V3 source as a backup.
with open(BACKUP, "x", encoding="utf-8") as f:
    f.write(original)

# Write modified V3 atomically.
temp_path = V3.with_suffix(".py.tmp")

try:
    with open(temp_path, "x", encoding="utf-8") as f:
        f.write(patched)

    os.replace(temp_path, V3)

finally:
    if temp_path.exists():
        temp_path.unlink()

# Verify the modified script and backup.
ast.parse(V3.read_text(encoding="utf-8"))

backup_hash = hashlib.sha256(BACKUP.read_bytes()).hexdigest()
new_hash = hashlib.sha256(V3.read_bytes()).hexdigest()

assert backup_hash == original_hash
assert new_hash != original_hash
assert "STRICT V3 CHECKPOINT VALIDATION" in V3.read_text()

print("=" * 70)
print("V3 RESUME SAFEGUARDS ADDED")
print("=" * 70)
print("V3 syntax: PASS")
print("V3 backup created:", BACKUP.exists())
print("Backup matches original V3:", backup_hash == original_hash)
print("V3 modified:", new_hash != original_hash)
print("V1/V2 modified: NO")
print("Training performed: NO")
print("Checkpoints created: NO")
print("Test split accessed: NO")
print("\nV3 SHA-256:", new_hash)
```

```python
# RECOVERY CHECK — inspect saved training artifacts only
# No training, no model updates, no test-set access.

from pathlib import Path
import json
import torch

EXP = Path(
    "/content/drive/MyDrive/Fruit_Freshness_Research/"
    "experiments/baseline_mobilenetv3_small"
)

print("=" * 65)
print("BASELINE TRAINING RECOVERY CHECK")
print("=" * 65)

for filename in [
    "latest_checkpoint.pt",
    "best_model.pt",
    "training_metrics.csv",
    "validation_metrics.csv",
    "per_class_metrics.csv",
    "run_status_v3.json",
]:
    path = EXP / filename
    if path.exists():
        print(f"FOUND  | {filename} | {path.stat().st_size:,} bytes")
    else:
        print(f"MISSING| {filename}")

# Read run status if available
status_file = EXP / "run_status_v3.json"
if status_file.exists():
    print("\n--- Run status ---")
    try:
        print(json.dumps(json.loads(status_file.read_text()), indent=2))
    except Exception as e:
        print("Could not read status JSON:", repr(e))

# Inspect latest checkpoint metadata only
latest = EXP / "latest_checkpoint.pt"

if latest.exists():
    print("\n--- Latest checkpoint metadata ---")
    try:
        ckpt = torch.load(latest, map_location="cpu", weights_only=False)

        for key in ckpt:
            if key in [
                "script_version",
                "completed_epoch",
                "best_macro_f1",
                "epochs_without_improvement",
                "history",
                "per_class_history",
            ]:
                value = ckpt[key]

                if key in ["history", "per_class_history"]:
                    print(f"{key}: {len(value)} records")
                    if key == "history":
                        print("Last recorded epochs:")
                        for row in value[-3:]:
                            print(row)
                else:
                    print(f"{key}: {value}")

    except Exception as e:
        print("Checkpoint inspection failed:", repr(e))

print("\nTraining launched by this cell: NO")
print("Test data accessed: NO")
print("=" * 65)
```
